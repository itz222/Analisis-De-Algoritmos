- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  
El Principio de Invarianza establece que dos implementaciones distintas de un mismo algoritmo no van a diferir en su tiempo de ejecucion mas que en una constante multiplicativa. Esto significa que el tiempo para resolver un problema depende de la naturaleza del algoritmo empleado y no de su implementacion (es decir, es independiente de factores externos como el tipo de computadora, el compilador o el lenguaje de programacion)
Definicion formal: Si un algoritmo tiene dos implementaciones distintas (I1 e I2) con tiempos de ejecucion T1(n) y T2(n) respectivamente, existen dos constantes reales c > 0 y n0 tales que para todo tamaño de entrada n >= n0 se cumple:
T1(n) <= c * T2(n)
Conclusiones clave del principio:
Independencia de la plataforma: El analisis asintotico permite estudiar la eficiencia de un algoritmo de forma general, sabiendo que las diferencias de hardware o codigo solo afectan el tiempo de ejecucion mediante un factor constante
Mismo orden de complejidad: Dos implementaciones "razonables" de un mismo algoritmo siempre perteneceran al mismo orden de complejidad (por ejemplo, si una es de orden cuadratico, la otra tambien lo sera)
. La diferencia de rendimiento entre ambas se reduce a una constante multiplicativa, comunmente llamada constante oculta


El Principio de Invarianza dice algo muy simple:
El tiempo de ejecucion de dos formas distintas de escribir un mismo algoritmo solo se diferencia por una constante multiplicativa (es decir, por un factor fijo de velocidad)
.
En la practica, esto significa:
No depende de la computadora: La eficiencia real de un algoritmo depende de su propio diseño y no de si usas una maquina mas rapida, un lenguaje de programacion diferente o tu habilidad al programar
.
Mismo ritmo de crecimiento: Si una forma de escribir el algoritmo es de orden cuadratico (como O(n^2)), cualquier otra forma razonable de escribir el mismo algoritmo tambien crecera al mismo ritmo cuadratico ante datos muy grandes
. Solo cambiara la velocidad base, conocida como la "constante oculta".

- *Explica cómo se utilizan las funciones de piso (floor) y techo (ceiling) en computación y detalla por qué la función exponencial positiva crece más rápido que cualquier polinomio según el texto.*  
**RESPUESTA:**

1. Uso de las funciones Piso (Floor) y Techo (Ceiling) en computación
Estas funciones se utilizan en matemáticas y computación para convertir números con decimales (reales) en números enteros de una forma completamente controlada
:
Función Piso (floor): Redondea un número hacia abajo, dándote el entero más grande que sea menor o igual que ese número
.
Ejemplo: El piso de 2.9 es 2
.
Función Techo (ceiling): Redondea un número hacia arriba, dándote el entero más pequeño que sea mayor o igual que ese número
.
Ejemplo: El techo de 6.1 es 7
.
Principales usos en la computación:
Dividir datos a la mitad: En algoritmos de búsqueda (como la búsqueda binaria) o estrategias de "divide y vencerás", constantemente se dividen conjuntos de datos a la mitad
. Si la cantidad de datos es impar (por ejemplo, 5 elementos), no puedes tener dos mitades iguales. En este caso, una mitad se calcula con el "piso" y la otra con el "techo", cumpliendo la propiedad exacta de que: Techo(n/2) + Piso(n/2) = n
.
Calcular bits para guardar información: Si tienes una colección de "n" objetos y quieres saber cuántos bits de memoria necesitas como mínimo para codificarlos (darle un identificador único a cada uno), la respuesta exacta se calcula usando el techo de un logaritmo: Techo(log2 n)
.
Establecer peores casos y límites: Al estimar el número máximo de pasos que dará un algoritmo recursivo, a menudo se calcula el entero inmediatamente superior (techo) de operaciones matemáticas para asegurar una cota superior segura
.

¿Por qué la función exponencial crece muchísimo más rápido que un polinomio?
Para entenderlo de forma fácil, mira cómo se construyen:
Polinomio (ejemplo: n^3): La base cambia, pero el exponente se queda fijo. Por ejemplo: 2^3, 3^3, 4^3...
Exponencial (ejemplo: 2^n): La base se queda fija, pero el exponente es el que crece. Por ejemplo: 2^2, 2^3, 2^4...
La razón del crecimiento explosivo: En la exponencial (2^n), cada vez que sumas 1 al valor de "n", multiplicas todo el resultado anterior por la base (en este caso, duplicas el valor completo)
. Esta multiplicación acumulada hace que el número crezca de forma salvaje. En cambio, en un polinomio el crecimiento es mucho más lento porque el exponente nunca cambia.
Matemáticamente, cuando los datos (n) se vuelven muy grandes, la división de cualquier polinomio entre una exponencial da cero
. Esto demuestra que la exponencial siempre "gana" y domina por completo al polinomio
.
Un ejemplo para notar la diferencia: Si tienes un grupo de 100 datos (n = 100):
Un algoritmo polinomial de n^3 necesita 1,000,000 (un millón) de operaciones
. Una computadora común hace esto en un milisegundo.
Un algoritmo exponencial de 2^n necesita tantas operaciones que, incluso usando una supercomputadora que haga un billón (10^12) de operaciones por segundo, tardaría unos 40,000,000,000 (40 mil millones) de años en terminar


Por esta razón, en la computación los algoritmos exponenciales se consideran intratables (inútiles en la vida real), mientras que con los polinomiales sí se puede trabajar
.

- *Define qué es un 'algoritmo', asegurándote de mencionar sus características clave, y describe qué significa el 'Problema de Paro'formulado por Alan Turing.*- *Sintetiza la diferencia entre realizar un 'Análisis a Priori' y una 'Prueba a Posteriori' para medir la eficiencia de un algoritmo.*  
**RESPUESTA:**

Un algoritmo es un conjunto finito de instrucciones precisas y ordenadas que se siguen paso a paso para resolver un problema o realizar una tarea
. Sus caracteristicas mas importantes son:
Finitud: Debe terminar siempre despues de un numero finito de pasos
.
Precision: Cada paso debe estar indicado sin ambiguedades, sabiendo exactamente que accion ejecutar a continuacion
.
Entrada y Salida: Tiene definidas las entradas que necesita para iniciar y produce un resultado correcto como salida
.
Determinismo: Si se ejecuta varias veces con el mismo conjunto de datos de entrada, siempre debe producir exactamente los mismos resultados
.
Efectividad: Cada paso o accion debe poder realizarse de forma sencilla en un tiempo finito


El Problema de Paro (Halting Problem), formulado por el matematico Alan Turing en 1936, consiste en intentar diseñar un algoritmo general que pueda determinar si cualquier otro programa de computadora, al recibir una entrada cualquiera, terminara su ejecucion en algun momento (se detendra) o si se quedara ejecutandose para siempre en un ciclo infinito.

Turing demostro matematicamente que este problema es indecidible
. Esto significa que es imposible que exista un algoritmo o programa de computadora general capaz de predecir correctamente en el 100% de los casos si un programa se detendra o no.

Analisis a Priori vs. Prueba a Posteriori
Para medir que tan eficiente es un algoritmo (cuanto tiempo y memoria consume), se utilizan dos enfoques:

Analisis a Priori (Teorico): Se realiza en la etapa de diseño, antes de escribir el codigo
. Consiste en obtener una funcion matematica que estima los limites de tiempo o recursos que necesitara el algoritmo segun el tamaño de la entrada de datos
. Su gran ventaja es que no depende de factores externos como el tipo de computadora, el lenguaje de programacion, el compilador o la habilidad del programador.

Prueba a Posteriori (Experimental): Se realiza ejecutando el programa ya programado
. Consiste en correr el algoritmo en una computadora real con datos de prueba especificos y medir fisicamente el tiempo y la memoria que consume
. Su gran desventaja es que depende por completo del hardware, el lenguaje, el compilador y los datos de prueba concretos utilizados durante el experimento.


- *Explica en qué consisten el peor caso, mejor caso y caso promedio al momento de evaluar el comportamiento de un algoritmo ante distintas entradas.*  
**RESPUESTA:**

Al evaluar cómo se comporta un algoritmo frente a diferentes entradas de datos, se analizan tres escenarios principales para medir su eficiencia de tiempo o de instrucciones:
1. El Peor Caso (Worst Case)
En qué consiste: Es la situación en la cual el algoritmo consume la máxima cantidad de tiempo o de instrucciones para resolver un problema de un tamaño determinado
.
Por qué importa: Funciona como una garantía absoluta o cota superior, asegurando que el programa nunca tardará más de ese límite, sin importar qué tan desfavorables sean los datos de entrada
. Al ser el análisis más seguro y conservador, es el que se estudia con mayor frecuencia en la computación
.
2. El Mejor Caso (Best Case)
En qué consiste: Es la situación ideal en la que el algoritmo realiza la menor cantidad de operaciones o toma el menor tiempo posible para completar su tarea
.
Por qué importa: Aunque es poco común que ocurra en la práctica, define el límite óptimo del algoritmo
. Si incluso en su mejor caso el rendimiento del algoritmo resulta lento o insatisfactorio, se puede descartar inmediatamente del diseño sin perder tiempo en análisis adicionales
.
3. El Caso Promedio (Average Case)
En qué consiste: Es el comportamiento esperado o típico del algoritmo cuando recibe datos aleatorios o comunes del mundo real
.
Por qué importa: Representa la medida de rendimiento más realista y útil para el uso cotidiano del programa
. Sin embargo, es el análisis más difícil de calcular matemáticamente, ya que requiere conocer o asumir de antemano la distribución de probabilidad de los datos de entrada

### Instrucciones:
1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
#FACTORIAL RECURSIVO
def factorial(n):
    if n <= 1:
        return 1  # Caso base: el factorial de 0 y 1 es 1
    return n * factorial(n - 1)  # Caso recursivo: n * (n-1)!
````

````python
#FIBONACCI ITERATIVO
def fibonacci(n):
    if n <= 1:
        return n  # Retorna 0 para n=0, y 1 para n=1

    anterior, actual = 0, 1
    for _ in range(2, n + 1):
        # Actualiza los valores sumando los dos números anteriores
        anterior, actual = actual, anterior + actual
    return actual
````

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
# --- PARTE 1: DEFINIR LAS FUNCIONES ---

def fibonacci_recursivo(n):
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)

def fibonacci_iterativo(n):
    if n <= 1:
        return n
    anterior, actual = 0, 1
    for _ in range(2, n + 1):
        anterior, actual = actual, anterior + actual
    return actual


# --- PARTE 2: MEDIR LOS TIEMPOS ---

print("Midiendo el tiempo del método iterativo:")
%timeit fibonacci_iterativo(35)

print("\nMidiendo el tiempo del método recursivo:")
%timeit fibonacci_recursivo(35)
````
